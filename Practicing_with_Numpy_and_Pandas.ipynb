## Super Cool Toolkits for Data! (NumPy and Pandas)

Hey little explorers! Imagine you have a big toy box, and inside are lots of different tools. Some tools help you build tall towers, and other tools help you sort your toy cars. In computer land, we have special 'toolkits' called **libraries** or **packages**. They are filled with helpful little robots that make working with numbers and tables super easy!

### What is NumPy? (The Super Number Robot!)

**NumPy** is like a super-duper number robot! If you have a long list of numbers, like `1, 2, 3, 4, 5`, NumPy can do amazing math tricks with them really, really fast!

*   **Imagine your number beads:** `[1, 2, 3, 4, 5]`
*   NumPy can quickly add them all up, or make each number much bigger, or even find the 'middle' number, faster than you can say 'supercalifragilisticexpialidocious'!
*   It's like the strong helper for all other number robots!

### What is Pandas? (The Smart Toy Organizer!)

**Pandas** is like a super smart toy organizer! Imagine your toys are all in a big table, just like a drawing pad with rows and columns, like this:

| Name | Age | Favorite Color |
|---|---|---|
| Timmy | 5 | Blue |
| Lily | 6 | Red |
| Max | 5 | Green |

*   Pandas helps you take your toy lists (like from a big book of toys) and put them neatly into a special table called a **DataFrame**.
*   With this special table, Pandas can help you:
    *   **Pick out just one kind of toy** (like only the 'Age' column).
    *   **Find toys that are special** (like only kids who are 5 years old).
    *   **Fix messy toy names** (like if a toy's name is spelled wrong).
    *   **Group toys together** (like putting all the 'blue' toys together and counting them).

So, remember:
*   **NumPy is for super-fast math with lists of numbers.**
*   **Pandas is for organizing, cleaning, and playing with your data tables like a pro!**

They are the best friends for playing with data!

```python
import pandas as pd
import numpy as np

# 1. Label-based selection with .loc (inclusive of row end bound '5')
print("--- .loc Selection ---")
print(df.loc[2:5, ['longitude', 'latitude']]) # Using existing columns

# 2. Position-based selection with .iloc (exclusive of row end bound '6', columns 1 and 3)
print("\n--- .iloc Selection ---")
print(df.iloc[2:6, [0, 1]]) # Using existing column indices for 'longitude' and 'latitude'

# 3. Set a specific column as the DataFrame index
print("\n--- Set Index to 'longitude' ---")
df_indexed = df.set_index('longitude') # Using an existing column
print(df_indexed.head())

# 4. Reset the index back to standard row integers (dropping the custom index column)
print("\n--- Reset Index (drop=True) ---")
df_reset = df_indexed.reset_index(drop=True)
print(df_reset.head())
```

Okay, little builders, let's learn some more tricks to play with our house data table!

1.  **`.loc` for Label-Based Selection (Picking by Name!)**:
    *   `df.loc[2:5, ['longitude', 'latitude']]` is like picking out specific toys from your toy box. You tell your helper robot, "Please get me the toys from box number 2 all the way to box number 5, and only the 'long' toys and the 'wide' toys!"
    *   `2:5` means it will show you rows 2, 3, 4, and 5. (Important: `.loc` is super friendly and includes *both* the start and end number you ask for!)
    *   `['longitude', 'latitude']` means it only wants to look at the 'long' column and the 'wide' column.

2.  **`.iloc` for Position-Based Selection (Picking by Number in Line!)**:
    *   `df.iloc[2:6, [0, 1]]` is like picking toys from a line. You tell your robot, "Please get me the toys that are in the 2nd, 3rd, 4th, and 5th spots in line, and only look at the first toy and the second toy in their hands!"
    *   `2:6` means it will show you toys at positions 2, 3, 4, and 5. (Important: `.iloc` is a little shy and *doesn't* include the very last number you ask for!)
    *   `[0, 1]` means it picks the column at the 0 spot (the first one) and the column at the 1 spot (the second one).

3.  **`set_index()` to Change the Index (Giving Houses New Special Numbers!)**:
    *   `df_indexed = df.set_index('longitude')` is like giving every house a *new special number* based on its 'long' address instead of just the usual 1, 2, 3. Now, the 'long' address becomes the super important number for finding each house!
    *   When you print it, you'll see the 'long' number moves to the very left, like a VIP ticket number for each house.

4.  **`reset_index(drop=True)` to Restore the Default Index (Back to Normal Numbers!)**:
    *   `df_reset = df_indexed.reset_index(drop=True)` is like saying, "Okay, let's take away those special 'long' numbers and just give our houses the regular 1, 2, 3 numbers again!"
    *   `drop=True` means that the old special 'long' number will disappear completely. Poof! If you didn't say `drop=True`, the 'long' number would just become a regular column again, not a special ticket number.

```python
# --- Setup: Creating proxy categorical columns for demonstration ---
# Create 'house_age_group' based on 'housing_median_age'
conditions_age = [
    df['housing_median_age'] < 15,
    (df['housing_median_age'] >= 15) & (df['housing_median_age'] < 30),
    df['housing_median_age'] >= 30
]
choices_age = ['New_Houses', 'Middle_Houses', 'Old_Houses']
df['house_age_group'] = np.select(conditions_age, choices_age, default='Unknown')

# Create 'population_density_group' based on 'population'
conditions_pop = [
    df['population'] < 500,
    (df['population'] >= 500) & (df['population'] < 2000),
    df['population'] >= 2000
]
choices_pop = ['Low_Density', 'Medium_Density', 'High_Density']
df['population_density_group'] = np.select(conditions_pop, choices_pop, default='Unknown')

print("Created 'house_age_group' and 'population_density_group' columns.")
print(df[['housing_median_age', 'house_age_group', 'population', 'population_density_group']].head())
```

```python
# --- Filtering & Querying Data ---

# Boolean Indexing & Conditional Filtering
print("\n--- Filtering: Single Condition (median_house_value > 300000) ---")
display(df[df['median_house_value'] > 300000].head())

print("\n--- Filtering: Multiple Conditions (house_age_group == 'Old_Houses' AND median_income > 5) ---")
display(df[(df['house_age_group'] == 'Old_Houses') & (df['median_income'] > 5)].head())

print("\n--- Filtering: Multiple Conditions (house_age_group == 'New_Houses' OR housing_median_age < 10) ---")
display(df[(df['house_age_group'] == 'New_Houses') | (df['housing_median_age'] < 10)].head())

# List Matching (isin)
print("\n--- Filtering: List Matching (house_age_group in ['New_Houses', 'Middle_Houses']) ---")
display(df[df['house_age_group'].isin(['New_Houses', 'Middle_Houses'])].head())

# Dynamic Querying (query)
print("\n--- Filtering: Dynamic Querying (house_age_group == 'Old_Houses' and median_income > 5) ---")
display(df.query("house_age_group == 'Old_Houses' and median_income > 5").head())

# Filtering Nulls (California housing dataset has no NaNs, so this will return an empty DataFrame)
# For demonstration, let's temporarily introduce a NaN to show how it works.
temp_df = df.copy()
temp_df.loc[0, 'total_bedrooms'] = np.nan
print("\n--- Filtering: Rows where 'total_bedrooms' is missing ---")
display(temp_df[temp_df['total_bedrooms'].isna()].head())
print("\n--- Filtering: Rows where 'total_bedrooms' is present (after temporary NaN) ---")
display(temp_df[temp_df['total_bedrooms'].notna()].head())
```

```python
# --- Vectorized Data Manipulation & Column Operations ---

# Adding & Modifying Columns
# Calculated Columns: Total property value estimate = median_house_value + (median_income * 100000)
# (This is a simplified example, not a real financial calculation)
print("\n--- Adding a Calculated Column ('Estimated_Property_Value') ---")
df['Estimated_Property_Value'] = df['median_house_value'] + (df['median_income'] * 100000)
display(df[['median_house_value', 'median_income', 'Estimated_Property_Value']].head())

# Conditional Logic with NumPy (np.where)
# Flag properties with high housing median age
print("\n--- Adding a Conditional Column ('Age_Flag') using np.where ---")
df['Age_Flag'] = np.where(df['housing_median_age'] >= 40, 'Very_Old', 'Standard_Age')
display(df[['housing_median_age', 'Age_Flag']].head())

# Multi-Condition Logic with NumPy (np.select)
print("\n--- Adding a Multi-Conditional Column ('Population_Category') using np.select ---")
conditions_population = [
    df['population'] < 700,
    (df['population'] >= 700) & (df['population'] < 2000),
    df['population'] >= 2000
]
choices_population = ['Small_Pop', 'Medium_Pop', 'Large_Pop']
df['Population_Category'] = np.select(conditions_population, choices_population, default='Unknown_Pop')
display(df[['population', 'Population_Category']].head())

# Sorting Data:
print("\n--- Sorting Data by 'house_age_group' (ascending) and 'median_house_value' (descending) ---")
display(df.sort_values(by=['house_age_group', 'median_house_value'], ascending=[True, False]).head())
```

```python
# --- Grouping & Aggregation ---

# GroupBy Operations
# Basic Aggregation: Average median_house_value by house_age_group
print("\n--- Basic Aggregation: Average median_house_value by house_age_group ---")
display(df.groupby('house_age_group')['median_house_value'].mean())

# Multiple Aggregations across Columns (agg)
print("\n--- Multiple Aggregations by 'house_age_group' ---")
display(df.groupby('house_age_group').agg(
    Avg_House_Value=('median_house_value', 'mean'),
    Max_House_Value=('median_house_value', 'max'),
    Total_Population=('population', 'sum'),
    Number_of_Districts=('longitude', 'count')
))

# Cross-Tabulation & Pivot Tables:
# Pivot table showing average median_house_value by house_age_group and Population_Category
print("\n--- Pivot Table: Average median_house_value by house_age_group and Population_Category ---")
display(df.pivot_table(index='house_age_group', columns='Population_Category', values='median_house_value', aggfunc='mean'))

# Frequency count across categorical variables
print("\n--- Cross-Tabulation: Counts of house_age_group by Population_Category ---")
display(pd.crosstab(df['house_age_group'], df['Population_Category']))
```

```python
# --- Pure NumPy Array Manipulation ---

# Convert Pandas series to NumPy arrays
print("\n--- Converting Pandas Series to NumPy Arrays ---")
house_values_np = df['median_house_value'].to_numpy()
house_ages_np = df['housing_median_age'].to_numpy()
print(f"Type of house_values_np: {type(house_values_np)}")
print(f"First 5 house values (NumPy array): {house_values_np[:5]}")

# Key NumPy Operations
print("\n--- NumPy Statistical Reductions ---")
print(f"Mean house value: {np.mean(house_values_np):.2f}")
print(f"Standard deviation of house values: {np.std(house_values_np):.2f}")
print(f"75th percentile of house values: {np.percentile(house_values_np, 75):.2f}")

# Array Searching & Filtering:
print("\n--- NumPy Array Searching & Filtering ---")
# Get indices where house value is greater than 400,000
indices_high_value = np.where(house_values_np > 400000)
print(f"Indices of houses with value > 400k (first 5): {indices_high_value[0][:5]}")

# Extract matching elements using boolean masking
high_value_houses = house_values_np[house_values_np > 400000]
print(f"High value houses (first 5): {high_value_houses[:5]}")

# Array Reshaping & Linear Algebra (example: log transformation)
print("\n--- NumPy Array Reshaping & Transformations (Log of house values) ---")
log_house_values = np.log(house_values_np)
print(f"Log of house values (first 5): {log_house_values[:5]}")
```

### **1. Filtering & Querying Data (Finding Specific Friends)**

Imagine you have a big group of friends, and you only want to play with the ones who like blue cars, or maybe the ones who are tall AND have red hats. That's what filtering is!

*   **Single Condition (`df[df['value'] > 300000]`)**: We asked our data table to show us only the houses that cost *more than* a certain big number. It's like saying, "Only show me the houses that are super expensive!"
*   **Multiple Conditions (`df[(condition1) & (condition2)]`)**: We asked to see houses that are *old* **AND** their owners earn *lots of money*. Or, houses that are *brand new* **OR** are *less than 10 years old*. It's like finding friends who like blue *and* green, or friends who like dogs *or* cats!
*   **List Matching (`df['group'].isin(['New', 'Middle'])`)**: We wanted to see houses that are either 'New' *or* 'Middle' aged. Like finding all the friends who like apples *or* bananas.
*   **Dynamic Querying (`df.query("condition")`)**: This is another way to ask the same questions, sometimes it's easier to read, like writing a special note to the data table.
*   **Filtering Nulls (`df[df['column'].isna()]`)**: Sometimes, some information is missing, like a puzzle piece. This helps us find the rows where a specific piece of information is missing (or where it's *not* missing).

### **2. Vectorized Data Manipulation & Column Operations (Giving New Hats or Changing Colors)**

This is like changing things for everyone in the group at once, without having to do it one by one!

*   **Calculated Columns (`df['New_Column'] = df['Old_Column'] + ...`)**: We made a brand new column that figured out an 'Estimated Property Value' by adding some numbers together from other columns. It's like giving everyone in your class a score for a game by adding their points from two different rounds.
*   **Conditional Logic (`np.where`)**: We gave houses a special 'Age Flag' if they were super old (over 40 years). It's like giving a gold star to kids who are taller than the swing set.
*   **Multi-Condition Logic (`np.select`)**: We sorted houses into 'Small Pop', 'Medium Pop', or 'Large Pop' categories based on how many people live there. This is like sorting your toys into 'small', 'medium', and 'big' boxes.
*   **Sorting Data (`df.sort_values(...)`)**: We lined up all the houses first by their age group (new to old) and then, for each age group, by how expensive they are (most expensive first!). It's like lining up kids first by their favorite color, and then by who's tallest in each color group.

### **3. Grouping & Aggregation (Counting Things in Groups)**

This is like putting all your similar toys together and then counting how many wheels each group has, or what the most popular color is in each group.

*   **Basic Aggregation (`df.groupby('group')['value'].mean()`)**: We put all the 'New Houses' together, all the 'Middle Houses' together, and all the 'Old Houses' together, and then asked, "What's the average price of a house in each group?"
*   **Multiple Aggregations (`df.groupby().agg(...)`)**: We asked for *lots* of information about each house age group: the average price, the highest price, the total number of people living there, and how many houses are in that group!
*   **Pivot Tables (`df.pivot_table(...)`)**: This is like making a special chart to see the average house price, but it's organized by both the house's age group *and* how many people live there (Small Pop, Medium Pop, Large Pop).
*   **Cross-Tabulation (`pd.crosstab(...)`)**: We counted how many houses fall into each combination of 'house age group' and 'population category'. It's like making a tally chart to see how many kids like blue and dogs, or red and cats.

### **4. Pure NumPy Array Manipulation (Super-Fast Math on Number Lists)**

NumPy is like a super-fast brain for just numbers! It's not so much about the labels and columns, but just crunching the numbers themselves.

*   **Converting to NumPy Arrays (`.to_numpy()`)**: We took our house prices and house ages from the Pandas table and turned them into simple, long lists of numbers that NumPy loves to work with.
*   **Statistical Reductions (`np.mean`, `np.std`, `np.percentile`)**: NumPy quickly told us the average house price, how spread out the prices are, and what price is higher than 75% of all other prices.
*   **Array Searching & Filtering (`np.where`)**: We quickly found out which houses had values over a really big number (like 400,000) and even got a list of those high prices.
*   **Array Reshaping & Linear Algebra (`np.log`)**: We did a special math trick (called a 'log transformation') on all the house prices at once, which can be useful for certain types of analysis. It's like doing a magic spell on all your numbers to change them in a specific way!
